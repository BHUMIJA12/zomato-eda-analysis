```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
import zipfile
import os

with zipfile.ZipFile('archive.zip', 'r') as z:
    z.extractall('.')
    print(z.namelist())
```

```python
df = pd.read_csv('Zomato-data-.csv')
df.head()
```

```python
print("Shape:", df.shape)
print("\nColumns:", df.columns.tolist())
print("\nMissing values:\n", df.isnull().sum())
```

```python
df['rate'].value_counts().head(15).plot(kind='bar', figsize=(10,4), color='tomato')
plt.title('Restaurant Ratings Distribution')
plt.xlabel('Rating')
plt.ylabel('Count')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
df['online_order'].value_counts().plot(kind='pie', autopct='%1.1f%%', figsize=(5,5), colors=['#ff6b6b','#4ecdc4'])
plt.title('Online Order Available?')
plt.ylabel('')
plt.show()
```

```python
df['listed_in(type)'].value_counts().head(10).plot(kind='barh', figsize=(8,4), color='steelblue')
plt.title('Top Restaurant Types')
plt.xlabel('Count')
plt.tight_layout()
plt.show()
```

```python
print("=== ZOMATO EDA SUMMARY ===")
print(f"Total Restaurants: {len(df)}")
print(f"Most Common Type: {df['listed_in(type)'].value_counts().index[0]}")
print(f"Online Order - Yes: {(df['online_order']=='Yes').sum()}")
print(f"Online Order - No: {(df['online_order']=='No').sum()}")
print(f"Average Votes: {df['votes'].mean():.0f}")
```

```python
sns.histplot(df['votes'], bins=20, kde=True, color='orange')
plt.title('Distribution of Restaurant Votes')
plt.xlabel('Votes')
plt.tight_layout()
plt.show()
```
